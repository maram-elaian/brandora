# 03 - Text Model Comparison (Full Scale)

##  Objective
This notebook scales the brand specification experiment from a single test case (Notebook 02) to the **full test set of 30 brand briefs**, generating a comprehensive comparison dataset across three language models.

---

## Methodology

### Input
- **Test Set:** 30 brand briefs from `data/test_briefs.json`
- **Models:** 3 LLMs (GPT-OSS-20B, GPT-OSS-120B, Qwen 3.8 27B)
- **Total Requests:** 30 × 3 = 90 API calls

### Process
Each brief is sent to each model with the **exact same prompt** to ensure fair comparison. Results are stored in a structured format with success/failure tracking.

### Output
- A JSON file containing all 90 results
- Saved to `evaluation/results/comparison_results.json`
- This file becomes the foundation for all subsequent evaluation

---

##  Research Questions This Notebook Answers

1. **Consistency:** Do models perform consistently across different industries?
2. **Failure Rate:** Which model fails most often under rate limits?
3. **Output Quality Patterns:** Do certain models excel at specific industries (e.g., tech vs. fashion)?

---

##  Technical Considerations

### Rate Limiting
Groq's free tier has request limits. We implement:
- A small delay (0.8s) between requests
- Error tracking (failed requests are logged, not lost)
- Progress bar for monitoring

### Data Structure
Each result entry contains:
- `brief_id`: Links back to the original test case
- `model`: Which model generated this output
- `status`: "success" or "failed"
- `output`: The JSON brand specification (if successful)
- `error`: Error message (if failed)
